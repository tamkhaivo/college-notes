```python
def gradient_descent(x, y, learning_rate=0.0001, iterations=1000000):
    """
    Performs gradient descent to find the best fit line (y = m*x + b).

    Args:
        x (np.array): The independent variable (e.g., hours studied).
        y (np.array): The dependent variable (e.g., test score).
        learning_rate (float): The step size for each iteration.
        iterations (int): The maximum number of iterations to perform.

    Returns:
        tuple: A tuple containing the final slope (m) and intercept (b).
    """
    # Start with some initial values for m and b
    m_curr = b_curr = 0
    n = len(x)
    # The learning rate is a critical hyperparameter. A value that is too large
    # can cause the cost to explode. 0.0001 is a safer starting point for this data.
    previous_cost = float('inf')

    print(f"Running Gradient Descent with learning rate = {learning_rate}...")
    print("-" * 50)

    for i in range(iterations):
        # Calculate the predicted y values with current m and b
        y_predicted = m_curr * x + b_curr
        
        # Calculate the cost (Mean Squared Error) using NumPy for performance.
        # This is much faster than using a Python loop or list comprehension.
        cost = np.mean((y - y_predicted)**2)

        # Stop when the cost reduction is negligible
        if math.isclose(previous_cost, cost, abs_tol=1e-20):
            print(f"\nStopping at iteration {i}. Costs have converged.")
            break
        
        previous_cost = cost

        # Calculate the partial derivatives (the gradients) using vectorized operations.
        # Derivative with respect to m
        md = -(2/n) * np.sum(x * (y - y_predicted))
        # Derivative with respect to b
        bd = -(2/n) * np.sum(y - y_predicted)

        # Update m and b by taking a step in the direction opposite to the gradient
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

        # Print progress periodically
        if i % 5000 == 0:
            print(f"Iteration {i:6d}: m = {m_curr:.4f}, b = {b_curr:.4f}, cost = {cost:.4f}")

    print("-" * 50)
    return m_curr, b_curr
```

```python
# 4. Create and train the Gradient Descent
X = df['math'].to_numpy()
y = df['cs'].to_numpy()

m_gd, b_gd = gradient_descent(X, y, learning_rate=0.0001)
```

```python
# 5. Display the results of the trained model
# The coefficients show the impact of each feature on the price.
print("\n--- Gradient Descent Results ---")
print(f"Slope (m): {m_gd:.6f}")
print(f"Intercept (b): {b_gd:.6f}")

# --- 2. Run Scikit-learn's Linear Regression ---
# Scikit-learn expects a 2D array for features, so we reshape x
model = LinearRegression()
model.fit(X.reshape(-1, 1), y)

m_sklearn = model.coef_[0]
b_sklearn = model.intercept_

print("\n--- Scikit-learn Results ---")
print(f"Coefficient (m): {m_sklearn:.6f}")
print(f"Intercept (b): {b_sklearn:.6f}")

# --- 3. Compare the results ---
print("\n--- Comparison ---")
print(f"The values for 'm' (slope) are very close: {m_gd:.6f} (GD) vs. {m_sklearn:.6f} (sklearn).")
print(f"The values for 'b' (intercept) are very close: {b_gd:.6f} (GD) vs. {b_sklearn:.6f} (sklearn).")
```
